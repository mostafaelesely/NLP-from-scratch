## ١. list: قايمة مرتبة وقابلة للتعديل

في AI: دفعة نصوص، قايمة tokens، قايمة نتايج. الترتيب مهم والتكرار مسموح.

```python
# قايمة جمل المستخدمين
texts = ["salah scored twice", "is rice injured", "next match for arsenal", "hello"]
# نضيف جملة جديدة في الآخر
texts.append("any transfer news")
# عدد الجمل
print(len(texts))
# أول وآخر عنصر
print(texts[0], "|", texts[-1])
# slicing: العناصر من 1 لحد قبل 3
print(texts[1:3])
# نقسم القايمة تدريب واختبار
train, test = texts[:4], texts[4:]
# حجم كل جزء
print(len(train), len(test))
```

```python
# نقطع الجملة لكلمات (tokens)
tokens = "is rice injured".split()
# نطبع الكلمات
print(tokens)
# comprehension: كل كلمة بحروف كبيرة
print([t.upper() for t in tokens])
# نرتب الجمل من الأقصر للأطول
print(sorted(texts, key=len))
```

## ٣. dict: مفتاح ← قيمة

أهم هيكل في AI: ملفات JSON، vocab (كلمة ← رقم)، وإعدادات الموديل.

```python
# قاموس نوايا الشات بوت
intents = {
# نية التحية
    "greeting": {"patterns": ["hello", "hi"], "reply": "Hello! Ask me about players."},
# نية الإصابات
    "injury": {"patterns": ["is he injured"], "reply": "Checking injury news..."},
# نقفل القاموس
}
# نوصل لقيمة جوه قاموس جوه قاموس
print(intents["injury"]["reply"])
# get: قيمة افتراضية لو المفتاح مش موجود
print(intents.get("transfer", "no such intent"))
# نضيف نية جديدة
intents["transfer"] = {"patterns": ["new signing"], "reply": "Latest transfers..."}
# أسماء النوايا
print(list(intents.keys()))
```

```python
# إعدادات الموديل (hyperparameters)
config = {"embedding_dim": 64, "lstm_units": 32, "dropout": 0.3, "epochs": 5}
# نقرأ إعداد
print(config["dropout"])
# نلف على المفتاح والقيمة
for k, v in config.items():
# نطبع الإعدادات بمحاذاة
    print(f"{k:14} = {v}")
```

```python
# قاموس الكلمات: أول رقمين لحشو ومجهول
vocab = {"<pad>": 0, "<unk>": 1}
# نلف على الجمل
for sent in texts:
# نلف على كلماتها
    for w in sent.split():
# لو الكلمة جديدة
        if w not in vocab:
# نديها رقم جديد
            vocab[w] = len(vocab)
# حجم الـ vocabulary
print(len(vocab))
# أول 8 كلمات
print({k: vocab[k] for k in list(vocab)[:8]})
# نحول جملة لأرقام والمجهول يبقى 1
ids = [vocab.get(w, vocab["<unk>"]) for w in "is salah injured today".split()]
# نطبع الأرقام
print(ids)
```

## ٤. set: عناصر فريدة

في AI: إزالة التكرار، stop words، وقياس التشابه بين جملتين.

```python
# كلمات الجملة الأولى كـ set
a = set("is rice injured today".split())
# كلمات الجملة التانية
b = set("is saka injured".split())
# التقاطع: الكلمات المشتركة
print(sorted(a & b))
# الاتحاد: كل الكلمات
print(sorted(a | b))
# الفرق: كلمات في الأولى بس
print(sorted(a - b))
# Jaccard similarity: المشترك ÷ الكل
print(round(len(a & b) / len(a | b), 2))
```

```python
# قايمة stop words
stop = {"is", "the", "a", "of"}
# جملة فيها كلمات شائعة
words = "is the news of the injury".split()
# نشيل الـ stop words
print([w for w in words if w not in stop])
# عدد الكلمات مقابل الفريدة
print(len(words), len(set(words)))
```

## ٥. Counter و defaultdict

بيوفروا كود كتير في العد والتجميع.

```python
# أدوات جاهزة للعد والتجميع
from collections import Counter, defaultdict
# كل كلمات الجمل (القديمة والجديدة) في قايمة واحدة
corpus = " ".join(texts + list(X)).split()
# نعد تكرار كل كلمة
cnt = Counter(corpus)
# أكتر 3 كلمات
print(cnt.most_common(3))
# كلمات ظهرت مرتين على الأقل
vocab_min2 = [w for w, c in cnt.items() if c >= 2]
# نطبعها
print(vocab_min2)
```

```python
# قاموس قيمه الافتراضية قايمة فاضية
by_intent = defaultdict(list)
# نلف على أزواج (جملة، نية)
for t, l in dataset:
# نجمع الجمل تحت نيتها
    by_intent[l].append(t)
# نطبع التجميع
print(dict(by_intent))
```

## ٦. deque: ذاكرة المحادثة

شات بوت محتاج يفتكر آخر كام رسالة بس. الـ deque بيمسح القديم لوحده.

```python
# قايمة من الطرفين بحجم أقصى
from collections import deque
# ذاكرة لآخر 3 رسائل بس
history = deque(maxlen=3)
# 4 رسائل متتابعة
for m in ["hi", "who is salah", "his club?", "his goals?"]:
# نضيف الرسالة
    history.append(m)
# الأقدم اتمسحت تلقائي
print(list(history))
```

## ٧. من قوايم لمصفوفة: padding

الموديل عايز كل الجمل بنفس الطول. نحوّل list of lists لمصفوفة NumPy.

```python
# جمل متحولة لأرقام بأطوال مختلفة
seqs = [[2, 3, 4], [5, 6], [7, 8, 9, 10]]
# أطول جملة
maxlen = max(len(s) for s in seqs)
# نكمل الناقص بأصفار (padding)
padded = np.array([s + [0] * (maxlen - len(s)) for s in seqs])
# المصفوفة النهائية
print(padded)
# الشكل: (عدد الجمل، أقصى طول)
print(padded.shape)
```
